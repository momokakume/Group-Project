```python
from astropy.table import Table
import matplotlib.pyplot as plt
```

```python
data = Table.read('Planet_lightcurve.fits', format='fits')
```

```python
data
```

```python
print(data.colnames)
print(data)

time = data['time [days]']
flux = data['flux']

plt.figure()
plt.plot(time, flux,'.')
plt.xlabel("Time")
plt.ylabel("Relative Flux")
```
